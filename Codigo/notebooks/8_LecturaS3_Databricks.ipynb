```python
%sql
WITH tablaFuente AS (
SELECT
    CONCAT(cliente,'-', dispositivo,'-Fase_', fase) AS codigoSerie
    ,voltaje
    ,corriente
    ,potencia
    ,factor_potencia
    ,frecuencia
    ,energia_activa
FROM tablaMediciones
)

SELECT
  codigoSerie
  ,count(1) AS cantidadRegistros
FROM tablafuente
GROUP BY codigoSerie
ORDER BY cantidadRegistros DESC
```
